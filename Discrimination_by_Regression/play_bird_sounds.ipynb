```python
import IPython
import os
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.image as mpimg
```

```python
files_train = np.genfromtxt("bird_sounds_files_train.csv", delimiter = ",", dtype = "str")
files_test = np.genfromtxt("bird_sounds_files_test.csv", delimiter = ",", dtype = "str")
```

```python
print("----------------------------------------------------------------")
print("****************************************************************")
print("----------------------------------------------------------------")
for i in [0, 2, 6]:
    sound_file = files_train[i]
    label = sound_file.split("_")[0]
    IPython.display.display(IPython.display.Audio(sound_file))

    img = mpimg.imread(label + ".jpg")
    plt.imshow(img)
    plt.axis("off")
    plt.show()

    print("----------------------------------------------------------------")
    print("****************************************************************")
    print("----------------------------------------------------------------")
```
